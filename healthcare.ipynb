```python
import pandas as pd
```

```python
df=pd.read_csv(r"C:\Users\leela\Desktop\Excel\healthcare_dataset\healthcare_dataset.csv")
```

```python
df
```

```python
new_cols={'Blood Type':'Blood_Type','Medical Condition':'Medical_Condition','Date of Admission':'Admission_Date',
        'Insurance Provider':'Insurance_Provider','Billing Amount':'Billing_Amount','Admission Type':'Admission_Type',
        'Discharge Date':'Discharge_Date','Test Results':'Test_Results'}
df.rename(columns=new_cols,inplace=True)
```

```python
df=df.drop_duplicates().reset_index(drop=True)
```

```python
df['Name']=df['Name'].str.strip().str.title()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
df['Admission_Date']=pd.to_datetime(df['Admission_Date'])
df['Discharge_Date']=pd.to_datetime(df['Discharge_Date'])
```

```python
df.head(5)
```

```python
df['Billing_Amount']=df['Billing_Amount'].abs().round(2)
```

```python
df['Doctor']=(df['Doctor']
    .str.replace(r'^(Mr\.|Mrs\.|Ms\.|Miss|Dr\.)\s+','',regex=True)
    .str.replace(r'\s+(MD|DDS|DVM|PHD|Jr\.|Sr\.|II|III|IV)$','',regex=True)
    .str.strip())
```

```python
df['Hospital']=(df['Hospital']
    .str.replace(',','',regex=True)
    .str.strip()
    .str.replace(r'^and\s+','',regex=True)
    .str.replace(r'\s+and$','',regex=True)
    .str.strip())
```

```python
print((df['Discharge_Date']<df['Admission_Date']).sum())
```

```python
print(df['Age'].min(),df['Age'].max())
```

```python
for col in ['Gender','Blood_Type','Medical_Condition','Insurance_Provider','Admission_Type','Medication',
    'Test_Results']:
    print(df[col].value_counts(),'\n')
```

```python
df['Length_Of_Stay']=(df['Discharge_Date']-df['Admission_Date']).dt.days
df['Admission_Year']=df['Admission_Date'].dt.year
```

```python
df['Age_Group']=pd.cut(df['Age'],
    bins=[0,15,30,60,120],
    labels=['Child','Adult','Middle Aged','Old Age'],
    right=False)
```

```python
df['Age_Group'].value_counts()
```

```python
df=df.drop(columns=['Room Number','Age'])
```
